```python
import random

class Ambiente:
    def __init__(self):
      self.status = {
          'A': random.choice(['Limpo', 'Sujo']),
          'B': random.choice(['Limpo', 'Sujo']),
      }
      self.posicao_agente = random.choice(['A', 'B'])

    def perceber(self):
      return self.status[self.posicao_agente]

    def executar(self, acao):
      if acao == 'Aspirar':
        # If the current location is dirty, clean it
        if self.status[self.posicao_agente] == 'Sujo':
          self.status[self.posicao_agente] = 'Limpo'
      elif acao == 'MoverEsquerda':
        # If agent is at 'B', move to 'A' (move left)
        if self.posicao_agente == 'B':
          self.posicao_agente = 'A'
      elif acao == 'MoverDireita':
        # If agent is at 'A', move to 'B' (move right)
        if self.posicao_agente == 'A':
          self.posicao_agente = 'B'
```

```python
def agente_reativo_simples(percepcao):
  posicao, status = percepcao
  if status == 'Sujo':
    return 'Aspirar'
  elif posicao == 'A':
    return 'MoverDireita'
  elif posicao == 'B':
    return 'MoverEsquerda'
```

```python
ambiente =  Ambiente()
print('Estado inicial', ambiente.status, '| Agente em:', ambiente.posicao_agente)

for passo in range(1, 6):
  percepcao = (ambiente.posicao_agente, ambiente.perceber())
  acao = agente_reativo_simples(percepcao)
  ambiente.executar(acao)
  print(f'Passo {passo}: percebeu {percepcao} -> {acao} | novo estado: {ambiente.status} | Agente em: {ambiente.posicao_agente}')
```

```python
class AgenteComMemoria:
  def __init__(self):
    self.memoria = {'A' : None, 'B': None}

  def decidir(self, percepcao):
    posicao, status = percepcao
    # Atualiza a memória com a percepção atual
    self.memoria[posicao] = status

    if status == 'Sujo':
      return 'Aspirar'

    if self.memoria['A'] == 'Limpo' and self.memoria['B'] == 'Limpo':
      return 'Parar'

    if posicao == 'A':
      if self.memoria['B'] == 'Sujo' or self.memoria['B'] is None:
        return 'MoverDireita'
      else:
        return 'MoverDireita'
    elif posicao == 'B':
      if self.memoria['A'] == 'Sujo' or self.memoria['A'] is None:
        return 'MoverEsquerda'
      else:
        return 'MoverEsquerda'
    return 'Parar'
```

```python
ambiente = Ambiente()
agente_memoria = AgenteComMemoria()
print('Estado inicial', ambiente.status, '| Agente em:', ambiente.posicao_agente)

for passo in range(1 , 10):
   percepcao = (ambiente.posicao_agente, ambiente.perceber() )
   acao = agente_reativo_simples(percepcao)
   agente_memoria.decidir(percepcao)
   ambiente.executar(acao)
   print(f'Passo {passo}: percebeu {percepcao} -> {acao} | Agente em: {ambiente.posicao_agente}')
```
